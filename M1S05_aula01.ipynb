## Funções internas de listas

* No python, toda estrutura possui métodos internos que realizam algumas funções. No python, as listas também usufruem desse benefício.

* Alguns dos principais métodos internos das listas são:

1. append()
2. insert()
3. pop()
4. del()

<br>

Comando: append()

A função append() irá adicionar um item ao final da lista. O único argumento é o valor a ser inserido.

````python
lista = ['João', "Maria", "Fernando"]

print(lista)

# Adicionando um novo sujeito
lista.append("Mônica")

print(lista)
````

<br>
<br>
<br>

Comando: insert()

A função insert irá incluir um item na posição que você especificar. O primeiro argumento é índice e o segundo é o valor inserido.

````python
lista = ['João', "Maria", "Fernando"]

print(lista)

# Adicionando um novo sujeito na posição 0
lista.insert(0, "Mônica")

print(lista)
````

<br>
<br>
<br>

Comando: pop()

A função pop irá apagar um item da lista. Podemos passar o índice por parâmetro para especificarmos qual item desejamos apagar. Ou podemos apagar o último item da lista, desde que não venhamos a passar nenhum valor.

````python
lista = ['João', "Maria", "Fernando"]

print(lista)

# Removendo o primeiro valor da lista
lista.pop(0)

print(lista)

# Removendo o último valor da lista
lista.pop(-1)
# lista.pop()

print(lista)
````

### Atividade Guiada

Faça uma lista vazia e após isso adicione 5 elementos dentro dela utilizando o input.

1° alternativa

````python
lista_init = list()
# lista_init = []

print(lista_init)

# Adicionar 5 elementos
lista_init.append(1)
lista_init.append(2)
lista_init.append(3)
lista_init.append(4)
lista_init.append(5)

print(lista_init)
````

2° alternativa

````python
lista_init = list()

print(lista_init)

elementos_desejados = [1, 2, 3, 4, 5]

for item in elementos_desejados:
    lista_init.append(item)


print(lista_init)

````

3º alternativa 


O extend() recebe uma lista (ou qualquer iterável) e adiciona cada item dela ao final da sua lista original.

````python
lista_init = []

lista_init.extend([0, 1, 2, 3, 4]) 

print(lista_init)
````

4º alternativa 


+= Funciona de forma idêntica ao extend, concatenando os itens da nova lista à lista atual.

````python
lista_init = []

lista_init += [0, 1, 2, 3, 4]

print(lista_init)
````

#### Crie uma lista com 5 elementos e remova os últimos 3 elementos de dentro dela. 

1° alternativa

````python
lista_init = [1, 2, 3, 4, 5]

print(lista_init)


lista_init.pop(-1)
lista_init.pop(-2)
lista_init.pop(-3)


print(lista_init)
````

2° alternativa

````python
lista = [0, 1, 2, 3, 4]

print(lista)

# Podemos usar Slicing (lista do índice de início até o índice -3)
lista = lista[:-3]

print(lista)
````

3° alternativa

````python
lista = [0, 1, 2, 3, 4]

# lembre que quando definido dessa forma o comando pop apaga o último elemento da lista
for _ in range(3):
    lista.pop()

print(lista)
````

4° alternativa

O comando ```del``` deleta uma lista ou parte dela quando usamos as técnicas de slicing.


````python
lista = [0, 1, 2, 3, 4]

# o comando del pode ser usar para deletar uma lista
del lista[-3:]

print(lista)
````

````python
teste = [1,2,3]

print(teste)

del teste

print(teste)
````
