```python
def somar(value_1, value_2, value_3 = 10):
    print(value_1 + value_2 + value_3)


somar(1, 2)
```

```python
# Passagem de parâmetro posicional
def somar(value_1, value_2, value_3 = 10):
    print(value_1)


somar(1, 3)

# Passagem de parâmetro nomeados
def somar(value_1, value_2, value_3 = 10):
    print(value_1) 


somar(value_2 = 12, value_1 = 33)
```
