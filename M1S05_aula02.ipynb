# Tuplas
<br>

Como vimos anteriormente listas no python são incrivelmente poderosas e úteis, porém não são ideais para todos os casos.

Listas não são ideais para todos os casos devido a sua natureza mutável o que permite que alguém possa alterar, <br>
modificar e até mesmo remover um elemento de dentro da nossa lista.

Porém, o python não oferece somente uma maneira de armazenar mais de uma informação ao mesmo tempo, temos também as tuplas. 

---

#### TUPLAS - Mutabilidade

* **Lista**: É mutável. Você pode usar .append(), .pop() ou alterar um valor (lista[0] = "novo").

* **Tupla**: É imutável. Depois de criada, você não consegue adicionar, remover ou alterar nenhum item.

---

#### TUPLAS - Sintaxe

* Lista: Usa colchetes [1, 2, 3].

* Tupla: Usa parênteses (1, 2, 3)

----

#### TUPLAS - Exemplo de uso prático

* Lista: Usada para coleções de coisas que podem mudar, como uma lista de compras ou nomes de usuários.

* Tupla: Usada para dados que devem ficar protegidos ou que representam uma estrutura fixa, como coordenadas geográficas (latitude, longitude) ou os dias da semana.

** Um detalhe interessante é que como as tuplas não mudam, o Python as processa um pouco mais rápido e elas ocupam menos memória que as listas.

---


````python
# Definindo tupla
tupla = tuple()

# Definindo tupla de uma outra forma
# tupla = ()
    
print(tupla)
````

````python
tupla = (1, 5, 6, 8, 2, 6)
print(tupla)
````

````python
tupla = (1, 5, 6, 8, 2, 6)

# Selecionando um item com base no seu índice
print(tupla[2])
````

#### Iterando sobre uma tupla

Tuplas também são iteráveis, isso significa que podemos percorrer uma lista utilizando estruturas de repetição.


A iteração tem o objetivo de percorrer os elementos contidos numa lista. Estudamos, que o Python possui 2 instruções definidas para que sejamos capazes de trabalhar com Iteração, que são, a estrutura while e a estrutura for.

---

````python
tupla = ('João', "Maria", "Fernando")

for i in tupla:
    print(i)
````

````python
tupla = ('João', "Maria", "Fernando", "Maria")

for i in range(0, len(tupla)):
    print(f'{i} ====> {tupla[i]}')
````

# Dicionários




<br>

Dicionários são estruturas de dados parecidas com as listas, porém ele usa um tipo diferente de organização. 

Ao invés de usar a indexação que já conhecemos ele opta por ir por um caminho diferente conhecido como chave e valor.

Para criarmos um dicionário é bem simples basta criarmos nossa variável e em seguida abrir o par de {}.

---

````python
# definindo dicionário
dicionario = {}

# Uma outra forma de definir dicionário
# dicionario = dict()


print(dicionario)

print(type(dicionario))
````

* Quando usamos os dicionários deixamos de seguir aquela ideia de indexação e agora começamos a fazer associações. 

* Iremos agora associar determinadas chaves a um valor específico.


````python
dict_pessoas = {
    "nome": "Maria"
}

print(dict_pessoas)
````

<br>

#### Dicionários: acessando valores


Podemos acessar os valores dele de forma similar a como acessamos os valores de uma lista, por exemplo, com a diferença de que usamos as chaves que definimos no lugar dos índices numéricos:


<br>

````python
# Criação de dicionário
dict_pessoas = {
    "Nome": "Maria",
    "Idade": 16,
    "Curso": "MEDICINA"
}

# Acessando um valor do diicionário
print(dict_pessoas["Nome"])
````

### Adicionando valores em um dicioário

Podemos adicionar valores dentro de um dicionário simplesmente criando uma chave que não existe dentro do mesmo e associando um valor a ele.

````python
# Criando dicionário vazio
turma = {}

# Adicionando novos elementos
turma["turno"] = "manhã"
turma["numero"] = 123


print(turma)
````

### modificando valores em um dicioário

Para modificar um valor já existente em um dicionário basta apenas chamar a sua chave e alterar o valor definido anteriormente.

````python
login = {
    "nome": "Felipe",
    "senha": 123456
}

print(login)

login["senha"] = "@dm123"

print(login)
````

#### Funções Internas dos dicionários

No python, toda estrutura possui métodos internos que realizam algumas funções. No python, as listas também usufruem desse benefício.


Comandos:

* ```clear()```

* ```copy()``` 

* ```items()```

* ```keys()```

<br>

#### copy()

Cópia todos os valores de um dicionario para outro.

````python
dicionario = {
    "nome": "Roberto",
    "idade": 42    
}

novo_dicionario = dicionario.copy()

print(novo_dicionario)
````

<br>

#### clear()

Remove todos os elementos de um dicionário.

````python
dicionario = {
    "nome": "Roberto",
    "idade": 42    
}

print(dicionario)


dicionario.clear()

print(dicionario)
````

<br>

#### keys()

Retorna somente todas as chaves de um dicionário.

````python
dicionario = {
    "nome": "Lopes",
    "CPF": "12345678",
    "RG": "1234567"
}

print(dicionario.keys())
````

<br>

#### items()

Devolve os dois elementos: chave e valor.

````python
dicionario = {
    "nome": "Lopes",
    "CPF": "12345678",
    "RG": "1234567"
}

print(dicionario.items())
````

<br>
<br>

#### Iterando sobre dicionários

````python
dicionario = {
    "nome": "Lopes",
    "CPF": "12345678",
    "RG": "1234567"
}

for chave in dicionario:
    print(chave)
````

````python
# Usando o comando values()

for valor in dicionario.values():
    print(valor)
````

````python
# Usando o comando items que vimos

for chave, valor in dicionario.items():
    print(f"O campo {chave} tem o valor: {valor}")
````

#### Estruturas compostas 

Uma das coisas mais interessantes é que podemos combinar muitas dessas estruturas, para expandir nossa capacidade de resolver problemas.
Essas estruturas podem ser úteis em vários contextos.

Vejamos uma que é muito comum: lista de dicionários

````python
lista_dicionario = [
    {"nome": "Ana", "idade": 25, "estado": "São Paulo"},
    {"nome": "Felipe", "idade": 33, "estado": "Santa Cataria"},
    {"nome": "Arnaldo", "idade": 56, "estado": "Rio de Janeiro"},
]

print(lista_dicionario)
````

````python
# Posso querer filtrar só o elemento 1
print(lista_dicionario[1])

# Mas e se eu quiser ver só o nome desse sujeito
print(lista_dicionario[1]["nome"])
````

Podemos iterar sobre essas estruturas compostas

````python
for pessoa in lista_dicionario:
    print(f"--- Registro de {pessoa['nome']} ---")

    for chave, valor in pessoa.items():
        print(f"{chave}: {valor}")
    print('\n')
````

Observe que esse foi só um exemplo, podemos ter listas dentro de listas, dicionarios dentro de dicionario, <br> 
lista dentro de dicionários, entre várias outras possibilidades. <br>
Existem outras estruturas em python, que vale a pena pesquisar um pouco mais (seujestão: sets).
