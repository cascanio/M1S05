Cria uma lista e adicione 5 elementos dentro dela usando input

```python
lista = []

lista.append(input("Digite seu nome: "))

lista.append(input("Qual sua idade?: "))

lista.append(input("Qual cidade você mora?: "))

lista.append(input("Qual a data de hoje?: "))

lista.append(input("Quantos filhos você tem?: "))


print (lista)
```

Crie uma lista com 5 elementos e remova os últimos 3 elementos de dentro dela.

```python
lista = [10, 20, 30, 40, 50]

lista = lista[:-3]

print(lista)
```

Faça uma tupla com 5 números e depois print todos eles em tela.

```python
tupla = (10, 20, 30, 40, 50)

print(tupla)

```

Crie uma tupla com 5 elementos e depois print o elemento que está no último index.

```python
tupla = (10, 20, 30, 40, 50)

print(tupla[-1])
```

Faça uma tupla com 5 elementos e depois percorra cada elemento da tupla com um for.

```python
tupla = (10, 20, 30, 40, 50)

for i in tupla:
    print(i)
```
